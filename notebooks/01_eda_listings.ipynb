# EDA — Listings Lima, Perú

Exploratory analysis of the silver layer before segmentation.

```python
import sys
sys.path.insert(0, '..')

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid', palette='muted')
```

## 1. Load silver layer

```python
df = pd.read_parquet('../data/silver/listings_silver.parquet')
print(f'Rows: {len(df):,} | Columns: {df.shape[1]}')
df.head()
```

## 2. Missing values

```python
missing = df.isnull().mean().sort_values(ascending=False)
missing[missing > 0].plot(kind='bar', figsize=(10, 4), title='Missing value rate')
plt.tight_layout()
plt.show()
```

## 3. Price distribution

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
df['price'].plot(kind='hist', bins=60, ax=axes[0], title='Price (raw)')
df['price'].apply(lambda x: x**0.5).plot(kind='hist', bins=60, ax=axes[1], title='Price (sqrt)')
plt.tight_layout()
plt.show()
df['price'].describe().round(2)
```

## 4. Room type breakdown

```python
df['room_type'].value_counts().plot(kind='bar', figsize=(8, 4), title='Room type')
plt.tight_layout()
plt.show()
```

## 5. Review score distribution

```python
df['review_scores_rating'].plot(kind='hist', bins=40, figsize=(8, 4), title='Review scores rating')
plt.tight_layout()
plt.show()
```

## 6. Top neighbourhoods by listing count

```python
top_n = df['neighbourhood_cleansed'].value_counts().head(15)
top_n.plot(kind='barh', figsize=(8, 5), title='Top 15 neighbourhoods')
plt.tight_layout()
plt.show()
```

## 7. Correlation heatmap

```python
num_cols = ['price', 'minimum_nights', 'number_of_reviews',
            'review_scores_rating', 'availability_365',
            'calculated_host_listings_count']
corr = df[num_cols].corr()
plt.figure(figsize=(8, 6))
sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm', vmin=-1, vmax=1)
plt.title('Feature correlations')
plt.tight_layout()
plt.show()
```

## 8. Median price by neighbourhood (top 15)

```python
top15 = top_n.index.tolist()
median_price = (
    df[df['neighbourhood_cleansed'].isin(top15)]
    .groupby('neighbourhood_cleansed')['price']
    .median()
    .sort_values(ascending=False)
)
median_price.plot(kind='barh', figsize=(8, 5), title='Median price by neighbourhood')
plt.tight_layout()
plt.show()
```
